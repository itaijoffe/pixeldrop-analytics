## Step 6: Missing Values Check
`.isnull().sum()` counts missing values per column.
Like counting NULLs in SQL — we saw these numbers in Script 01!

```python
print('Missing values per table:\n')
for name, df in tables.items():
    nulls = df.isnull().sum()
    nulls = nulls[nulls > 0]  # only show columns that have nulls
    print(f'--- {name} ---')
    if len(nulls) == 0:
        print('  No missing values!')
    else:
        print(nulls.to_string())
    print()
```

## Step 8: Check Numeric Issues
Look for bad quantities, impossible discounts, and price format issues.

```python
# Check quantity column - convert to numeric first
order_items['quantity_check'] = pd.to_numeric(order_items['quantity'], errors='coerce')

bad_qty = order_items[order_items['quantity_check'] <= 0]
null_qty = order_items[order_items['quantity_check'].isnull()]

print(f'Bad quantities (<=0): {len(bad_qty)}')
print(f'Null quantities: {len(null_qty)}')

# Drop the temp column
order_items.drop(columns=['quantity_check'], inplace=True)
```

```python
# Check discount column for impossible values (>100)
order_items['discount_check'] = pd.to_numeric(order_items['discount_%'], errors='coerce')

bad_discount = order_items[order_items['discount_check'] > 100]
print(f'Impossible discounts (>100%): {len(bad_discount)}')
print(f'Sample bad discounts:')
print(bad_discount[['item_id', 'discount_%']].head())

order_items.drop(columns=['discount_check'], inplace=True)
```
